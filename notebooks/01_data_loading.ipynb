```python
import pandas as pd
import numpy as np

print("Libraries imported")
```

```python
df = pd.read_csv("../data/telco_churn.csv")

df.head()
```

```python
print("Shape (rows, columns):", df.shape)
print()
print("Column names:")
print(list(df.columns))
print()
df.info()
```

```python
print("Missing values per column:")
print(df.isnull().sum())
print()
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate customer IDs:", df["customerID"].duplicated().sum())
print()
print("Blank TotalCharges:", (df["TotalCharges"].str.strip() == "").sum())
```

```python
df_clean = df.copy()

df_clean["TotalCharges"] = pd.to_numeric(df_clean["TotalCharges"], errors="coerce")

print("Missing TotalCharges after conversion:", df_clean["TotalCharges"].isnull().sum())
print()
df_clean[df_clean["TotalCharges"].isnull()][["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]]
```

```python
df_clean["TotalCharges"] = df_clean["TotalCharges"].fillna(0)

print("Missing TotalCharges now:", df_clean["TotalCharges"].isnull().sum())
print("Data type now:", df_clean["TotalCharges"].dtype)
```

```python
before = len(df_clean)
df_clean = df_clean.drop_duplicates()
print("Rows removed:", before - len(df_clean))
print()

print(df_clean[["tenure", "MonthlyCharges", "TotalCharges"]].describe())
print()
print("SeniorCitizen values:", df_clean["SeniorCitizen"].unique())
print("Churn values:", df_clean["Churn"].unique())
```

```python
df_clean["Churn"] = df_clean["Churn"].map({"Yes": 1, "No": 0})

print(df_clean["Churn"].value_counts())
print()
print("Churn rate:", round(df_clean["Churn"].mean() * 100, 2), "%")

df_clean.to_csv("../data/telco_clean.csv", index=False)
print()
print("Saved. Shape:", df_clean.shape)
```
